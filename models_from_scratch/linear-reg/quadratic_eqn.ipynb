```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0,30,1)
x1 = x
x2 = x+7
# defining a known linearity
eqn =  x2**3 + x1 + 5 + np.random.normal(0, 1000, 30)
y = eqn
```

```python
x,y
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x1,y, label='x1')
# plt.scatter(x2,y, label='x2')
plt.title('Linear')
plt.legend()
plt.show()
```

```python
X  =np.column_stack([x1,x2])

from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X,y)
```

```python
y_pred = lr.predict(X)
```

```python
plt.figure(figsize=(10,6))
plt.plot(x,y, label ="true")
plt.scatter(x,y_pred, color='red', alpha =0.6, label='pred')
plt.title('Linear')
# plt.xlim(4,27)
# plt.xticks(np.arange(4,27))
plt.legend()
plt.show()
```

```python
lr.coef_, lr.intercept_
```

```python
from sklearn.metrics import root_mean_squared_error

root_mean_squared_error(y,y_pred)
```

```python
from sklearn.metrics import mean_absolute_error


mean_absolute_error(y,y_pred)
```

```python
plt.plot(x,(y-y_pred))
```

```python
from sklearn.metrics import r2_score

r2_score(y,y_pred)
```

**Observations**

1. R2 score tells us that our quadratic model "accounts for" 88% of the movement in our cubic data.
2. RMSE > MAE, tells that some values have high errors.
